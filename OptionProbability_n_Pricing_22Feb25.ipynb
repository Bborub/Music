1. Binomial Model
This script builds a binomial tree and calculates the probability of the stock price exceeding the strike at expiration.

```python
import numpy as np

# Inputs
S = 100  # Current stock price
K = 105  # Strike price
T = 30 / 365  # Time to expiration (30 days)
sigma = 0.20  # Volatility
r = 0.05  # Risk-free rate
n = 2  # Number of time steps (keep small for simplicity)

# Binomial parameters
dt = T / n
u = np.exp(sigma * np.sqrt(dt))  # Up factor
d = 1 / u  # Down factor
p = (np.exp(r * dt) - d) / (u - d)  # Risk-neutral probability

# Build price tree (just for final nodes)
final_prices = []
for i in range(n + 1):
    up_moves = i
    down_moves = n - i
    price = S * (u ** up_moves) * (d ** down_moves)
    final_prices.append(price)

# Calculate probability of ITM
itm_count = sum(1 for price in final_prices if price > K)
total_paths = 2 ** n
prob_itm = itm_count / total_paths

# More precise: use binomial probabilities
from scipy.stats import binom
prob_itm_binom = sum(binom.pmf(k, n, p) for k in range(n + 1) if S * (u ** k) * (d ** (n - k)) > K)

print(f"Final prices: {final_prices}")
print(f"Crude ITM probability: {prob_itm:.4f}")
print(f"Binomial ITM probability: {prob_itm_binom:.4f}")
```

Output Explanation:
final_prices: Possible stock prices at expiration (e.g., [$121, $99, $81] for 2 steps).
prob_itm: Fraction of paths exceeding the strike (crude estimate).
prob_itm_binom: Weighted by risk-neutral probabilities (more accurate).

2. Monte Carlo Simulation
This script simulates 10,000 stock price paths and estimates the ITM probability.

```python
import numpy as np

# Inputs
S = 100  # Current stock price
K = 105  # Strike price
T = 30 / 365  # Time to expiration (30 days)
sigma = 0.20  # Volatility
r = 0.05  # Risk-free rate
n_simulations = 10000  # Number of simulations

# Random numbers for simulation
np.random.seed(42)  # For reproducibility
z = np.random.normal(0, 1, n_simulations)

# Simulate final prices using geometric Brownian motion
drift = (r - 0.5 * sigma**2) * T
diffusion = sigma * np.sqrt(T) * z
final_prices = S * np.exp(drift + diffusion)

# Calculate ITM probability
itm_count = np.sum(final_prices > K)
prob_itm = itm_count / n_simulations

print(f"First 5 simulated prices: {final_prices[:5]}")
print(f"Monte Carlo ITM probability: {prob_itm:.4f}")
```

Output Explanation:
final_prices: Array of simulated prices at expiration.
prob_itm: Fraction of simulations where the price exceeds the strike.

```python
import numpy as np
from scipy.stats import norm

# Inputs
S = 100  # Current stock price
K = 105  # Strike price
T = 30 / 365  # Time to expiration (30 days)
sigma = 0.20  # Implied volatility (assumed from market data)
r = 0.05  # Risk-free rate

# Black-Scholes d1 and d2
d1 = (np.log(S / K) + (r + 0.5 * sigma**2) * T) / (sigma * np.sqrt(T))
d2 = d1 - sigma * np.sqrt(T)

# Probabilities
delta = norm.cdf(d1)  # Delta (approximate ITM probability)
prob_itm = norm.cdf(d2)  # N(d2) = exact risk-neutral ITM probability

print(f"d1: {d1:.4f}, d2: {d2:.4f}")
print(f"Delta (approx. ITM probability): {delta:.4f}")
print(f"N(d2) (exact ITM probability): {prob_itm:.4f}")
```
