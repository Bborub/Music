**Chi-Square Goodness of Fit test from Unit 15 PPT**

**Below is from Grok**

```python
# Import the necessary library
from scipy.stats import chisquare

# Observed frequencies from the table
observed = [17, 40, 68, 51, 24]

# Expected probabilities from the table
expected_prob = [0.1, 0.2, 0.4, 0.2, 0.1]

# Calculate total observed frequency
total_observed = sum(observed)

# Convert expected probabilities to expected frequencies
expected = [total_observed * prob for prob in expected_prob]

# Perform the Chi-Square Goodness of Fit test
chi_stat, p_value = chisquare(f_obs=observed, f_exp=expected)

# Print the results
print(f"Chi-Square Statistic: {chi_stat:.3f}")
print(f"P-value: {p_value:.3f}")

# Interpret the result (using a significance level of 0.05)
if p_value < 0.05:
    print("Reject the null hypothesis: The observed distribution does not fit the expected distribution.")
else:
    print("Fail to reject the null hypothesis: The observed distribution fits the expected distribution.")
```

```python
# Import the necessary library
from scipy.stats import chisquare

# Observed frequencies from the table
observed = [17, 40, 68, 51, 24]

# Expected probabilities from the table
expected_prob = [0.1, 0.2, 0.4, 0.2, 0.1]

# Calculate total observed frequency
total_observed = sum(observed)

# Convert expected probabilities to expected frequencies
expected = [total_observed * prob for prob in expected_prob]

# Calculate degrees of freedom (number of categories - 1)
df = len(observed) - 1

# Perform the Chi-Square Goodness of Fit test
chi_stat, p_value = chisquare(f_obs=observed, f_exp=expected)

# Print the results with p-value rounded to 3 decimal places and degrees of freedom
print(f"Chi-Square Statistic: {chi_stat:.3f}")
print(f"Degrees of Freedom: {df}")
print(f"P-value: {p_value:.3f}")

# Interpret the result (using a significance level of 0.05)
if p_value < 0.05:
    print("Reject the null hypothesis: The observed distribution does not fit the expected distribution.")
else:
    print("Fail to reject the null hypothesis: The observed distribution fits the expected distribution.")
```

```python
# Import the necessary library
from scipy.stats import chi2_contingency
import numpy as np

# Observed data in a 2D array (contingency table without totals)
observed = np.array([
    [24, 29, 23],  # Imported
    [15, 35, 21]   # Domestic
])

# Perform the Chi-Square Test for Homogeneity
chi_stat, p_value, df, expected = chi2_contingency(observed)

# Print the results with p-value rounded to 3 decimal places
print(f"Chi-Square Statistic: {chi_stat:.3f}")
print(f"Degrees of Freedom: {df}")
print(f"P-value: {p_value:.3f}")
print("\nExpected Frequencies:")
print(expected)

# Interpret the result (using a significance level of 0.05)
if p_value < 0.05:
    print("Reject the null hypothesis: The preference distributions are not the same for Imported and Domestic beers.")
else:
    print("Fail to reject the null hypothesis: The preference distributions are the same for Imported and Domestic beers.")
```

```python
# Import the necessary libraries
from scipy.stats import chi2_contingency
import numpy as np

# Observed data in a 2D array (contingency table without totals)
observed = np.array([
    [24, 29, 23],  # Imported
    [15, 35, 21]   # Domestic
])

# Perform the Chi-Square Test for Homogeneity
chi_stat, p_value, df, expected = chi2_contingency(observed)

# Calculate the Likelihood Ratio (G-test) statistic
# G = 2 * Σ(O * ln(O/E)) where O is observed and E is expected
g_stat = 0
for i in range(observed.shape[0]):
    for j in range(observed.shape[1]):
        if observed[i][j] > 0:  # Avoid log(0) issues
            g_stat += observed[i][j] * np.log(observed[i][j] / expected[i][j])
g_stat *= 2

# Print the results with p-value rounded to 3 decimal places
print(f"Chi-Square Statistic: {chi_stat:.3f}")
print(f"Likelihood Ratio (G-test) Statistic: {g_stat:.3f}")
print(f"Degrees of Freedom: {df}")
print(f"P-value (Chi-Square): {p_value:.3f}")

# Display expected frequencies with labels for clarity
print("\nExpected Frequencies (Rows: Imported, Domestic | Columns: Excellent, Average, Poor):")
print("Imported:", [f"{val:.3f}" for val in expected[0]])
print("Domestic:", [f"{val:.3f}" for val in expected[1]])

# Interpret the result (using a significance level of 0.05)
if p_value < 0.05:
    print("Reject the null hypothesis: The preference distributions are not the same for Imported and Domestic beers.")
else:
    print("Fail to reject the null hypothesis: The preference distributions are the same for Imported and Domestic beers.")
```
